```python
def zipper_file(filename, strings):
   with open(filename, 'w') as f:

       for i in range(0, len(strings), 2):
           string1 = strings[i]
           string2 = strings[i+1]

           zipped = ""
           for position in range(len(string1)):
               zipped += string1[position]
               zipped += string2[position]

           f.write(zipped + "\n")
```

```python
def reformat_date(dates_list):
   month_map = {
       "01": "January",
       "02": "February",
       "03": "March",
       "04": "April",
       "05": "May",
       "06": "June",
       "07": "July",
       "08": "August",
       "09": "September",
       "10": "October",
       "11": "November",
       "12": "December"
   }
   formatted_dates = []
   for date_str in dates_list:
       parts = date_str.split('/')
       month_num = parts[0]
       day = parts[1]
       year = parts[2]

       month_name = month_map[month_num]

       formatted_str = f"office hours on {month_name} {day}, {year}"
       formatted_dates.append(formatted_str)

   return formatted_dates

sample_dates = ["10/24/2026", "01/05/2025"]
print(reformat_date(sample_dates))


sample_words = ["cat", "dog", "apple", "juice"]
zipper_file("output.txt", sample_words)



```
