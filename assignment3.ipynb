```python
#Part A
#Dictionaries
#1. Create an empty dictionary
inventory = {}
```

```python
#2. Store the first product details in variables
ProductNol_name = "Wireless Headphones"
ProductNol_quantity = 8
ProductNol_price = 3500
ProductNol_releaseYear = 2023
```

```python
#3. Add the first product details to inventory
inventory ["ProductNol_name"] = ProductNol_name
inventory ["ProductNol_quantity"] = ProductNol_quantity
inventory ["ProductNol_price"] = ProductNol_price
inventory ["ProductNol_releaseYear"] = ProductNol_releaseYear
```

```python
#4. Store the second product details in variables
ProductNo2_name = "Smart Backpack"
ProductNo2_quantity = 12
ProductNo2_price = 2200
ProductNo2_releaseYear = 2024
```

```python
#5. Add the second product details to inventory
inventory ["ProductNo2_name"] = ProductNo2_name
inventory ["ProductNo2_quantity"] = ProductNo2_quantity
inventory ["ProductNo2_price"] = ProductNo2_price
inventory ["ProductNo2_releaseYear"] = ProductNo2_releaseYear
```

```python
#6. Display the products present in inventory
print("Products present in inventory:")
for product, details in inventory.items():
   print(product, ":", details)
```

```python
#7. Check the release years of both products
print("\nChecking release years:")
print("ProductNol_releaseYear in inventory:", "ProductNol_releaseYear" in inventory)
print("ProductNo2_releaseYear in inventory:", "ProductNo2_releaseYear" in inventory)
```

```python
#8. Delete release year of both products
del inventory ["ProductNol_releaseYear"]
del inventory ["ProductNo2_releaseYear"]
print("\nInventory after deleting release years:")
print(inventory)
```

```python
#Part B
#Tuples
#Create a tuple containing product prices
prices = (275, 450, 125, 625, 180, 390, 510)
```

```python
#Find and print the highest price
highest_price = max(prices)
print("Highest Price:", highest_price)
```

```python
#Find and print the lowest price
lowest_price = min(prices)
print("Lowest Price:", lowest_price)
```

```python
#3. Calculate total sum of all prices
total = sum(prices)
print("Total sum of all prices:", total)
```

```python
#4. Convert tuple into sorted list in ascending order
sorted_prices = sorted (prices)
print("Sorted list:", sorted_prices)
```

```python
#5. Trying to modify an element in the tuple
try:
   prices [0] = 500
except TypeError:
   print("Tuple cannot be modified because tuples are immutable.")
```
